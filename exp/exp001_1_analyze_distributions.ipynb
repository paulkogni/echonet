```python
import numpy as np 
import pandas as pd
import os
```

```python
path_to_csv = '/home/paul/Downloads/econet_ped/echonetpediatric/pediatric_echo_avi/pediatric_echo_avi/A4C/FileList.csv'
path_train = '/home/paul/Downloads/echonet_ped_preprocessed/TRAIN'
path_val = '/home/paul/Downloads/echonet_ped_preprocessed/VAL'
path_test = '/home/paul/Downloads/echonet_ped_preprocessed/TEST'
```

```python
df = pd.read_csv(path_to_csv)
df.head()
```

```python
def get_filenames_from_dir(path):
    return {
        f[:-3] + '.avi' for f in os.listdir(path)
        if os.path.isfile(os.path.join(path, f))
    }

train_files = get_filenames_from_dir(path_train)
val_files   = get_filenames_from_dir(path_val)
test_files  = get_filenames_from_dir(path_test)
```

```python
train_files
```

```python
df_train_used = df[df["FileName"].isin(train_files)]
df_val_used   = df[df["FileName"].isin(val_files)]
df_test_used  = df[df["FileName"].isin(test_files)]
```

```python
assert df_train_used["Split"].isin(range(0, 8)).all()
assert (df_val_used["Split"] == 8).all()
assert (df_test_used["Split"] == 9).all()
```

```python
def age_distribution(df_subset):
    return (
        df_subset
        .groupby("Age")
        .size()
        .reset_index(name="n_subjects")
        .rename(columns={"Age": "age"})
        .sort_values("age")
        .reset_index(drop=True)
    )

train_age_df = age_distribution(df_train_used)
val_age_df   = age_distribution(df_val_used)
test_age_df  = age_distribution(df_test_used)
```

```python
test_age_df
```

```python
train_age_df.to_csv('train_distribution.csv', index=False)
val_age_df.to_csv('val_distribution.csv', index=False)
test_age_df.to_csv('test_distribution.csv', index=False)
```
