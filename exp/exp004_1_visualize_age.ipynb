```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
list_of_agegroups = ['Babies (0)', 'Toddler (1-2)', 'Preschooler (3-4)', 'Gradeschooler (5-11)', 'Teens (12-18)']

list_of_settings = ['adults -> kids', 'kids -> kids', 'all -> kids']

adults_kids_dice = [0.7274, 0.8780, 0.8627, 0.8823, 0.8782]
adults_kids_std = [0.3145, 0.0562, 0.0802, 0.0567, 0.0788]

kids_kids_dice = [0.8517, 0.9226, 0.9164, 0.9116, 0.9101]
kids_kids_std = [0.1686, 0.0400, 0.0441, 0.0413, 0.0453]

all_kids_dice = [0.7989, 0.8905, 0.8934, 0.8874, 0.8821]
all_kids_std = [0.2230, 0.0422, 0.0407, 0.0594, 0.0690]
```

```python
plt.figure(figsize=(10, 6))

plt.errorbar(list_of_agegroups, adults_kids_dice, yerr=adults_kids_std, marker='o', linestyle='-', label='adults -> kids', capsize=4)
plt.errorbar(list_of_agegroups, kids_kids_dice, yerr=kids_kids_std, marker='o', linestyle='-', label='kids -> kids', capsize=4)
plt.errorbar(list_of_agegroups, all_kids_dice, yerr=all_kids_std, marker='o', linestyle='-', label='all -> kids', capsize=4)

plt.xlabel('Age Groups')
plt.ylabel('Dice Score')
plt.title('Segmentation Performance Across Age Groups')
plt.legend()
plt.grid(True)
plt.tight_layout()

plt.show()
```
